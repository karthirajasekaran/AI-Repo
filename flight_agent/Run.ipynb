```python
###### RAG embeddings - verification part 2 ##########
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_community.vectorstores import Chroma
from chromadb.utils import embedding_functions
import chromadb

client = chromadb.PersistentClient(
    path="dbfs:/user/hive/warehouse/krajasekaran.db/chroma_db_krajasekaran"
)

collection = client.get_collection("travel_policy_collection")

embeddings = HuggingFaceEmbeddings(model_name="all-MiniLM-L6-v2")
vector_store = Chroma(client=client,
                      collection_name="travel_policy_collection",
                      embedding_function=embeddings)

airlines = ["general", "delta air lines", "american airlines"]

for airline in airlines:
    docs = vector_store.similarity_search("baggage policy", k=3, filter={"airline_name": airline})
    print(f"\n--- Results for {airline} ---")
    for d in docs:
        print(d.page_content[:200])  # print first 200 chars
```

```python
###### Agent , tools and utils imports ##########
from langchain_core.messages import HumanMessage
from src.agent import agent_app  
from src.utils import get_input_with_timeout, run_agent_loop  
import langsmith
from transformers import logging
# This silences the "Some weights of the model checkpoint..." warnings
logging.set_verbosity_error()
```

```python
###### Running the agent ##########
import uuid
session_id = str(uuid.uuid4())
config = {"configurable": {"thread_id": session_id}}
for user_input in run_agent_loop(agent_app):
    try:
        for event in agent_app.stream(
            {"messages": [HumanMessage(content=user_input)]},config=config,
        ):
            if "agent" in event:
                msg = event["agent"]["messages"][-1]
                print(f"Agent: {msg.content}")

    except Exception as e:
        print(f"Error: {e}")
```
